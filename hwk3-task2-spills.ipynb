````python
# convert spill_date from string to datetime
spill_clean = spills.copy()
spill_clean['spill_date'] = pd.to_datetime(spill_clean['spill_date'])

# check spill_date head and type
print(spill_clean['spill_date'].dtype)
spill_clean['spill_date'].head(3)
````
